```python
# Step 2: Install package and ensure 'src' is in sys.path for the active kernel
!pip install -e . -q

import sys, site
from pathlib import Path

# Force refresh of site-packages
site.main()

# Ensure 'src' directory is in sys.path (for direct imports before reload)
# This handles Kaggle's caching where .pth files might not be refreshed
for candidate in [Path.cwd(), Path.cwd() / "usrecon", Path.cwd() / "usrecon-root", *Path.cwd().parents]:
    src_dir = candidate / "src"
    if (src_dir / "usrecon").exists():
        if str(src_dir) not in sys.path:
            sys.path.insert(0, str(src_dir))
        print(f"[usrecon] Added {src_dir} to sys.path")
        break

# Re-import usrecon to ensure fresh module load (fixes Kaggle cache issues)
if 'usrecon' in sys.modules:
    del sys.modules['usrecon']
for mod_name in list(sys.modules.keys()):
    if mod_name.startswith('usrecon.'):
        del sys.modules[mod_name]
```
